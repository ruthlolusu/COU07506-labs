```python
import numpy as np
import matplotlib.pyplot as plt
from scipy import signal
from scipy.io import wavfile

plt.rcParams["figure.figsize"] = (9, 3.5)
```

```python
n = np.arange(-5, 11)

delta = (n == 0).astype(float)
step = (n >= 0).astype(float)
ramp = n * step
expo = 0.8**n * step
```

```python
fig, axes = plt.subplots(1, 4, figsize=(13, 2.6))

for ax, x, title in zip(
    axes,
    [delta, step, ramp, expo],
    ["δ(n)", "u(n)", "r(n)", "(0.8)^n u(n)"]
):
    ax.stem(n, x, basefmt="k-")
    ax.set_title(title)
    ax.set_xlabel("n")

plt.tight_layout()
plt.show()
```

```python
n = np.arange(0, 21)

a_values = [0.5, -0.8, 1.1]

fig, axes = plt.subplots(1, 3, figsize=(12, 3))

for ax, a in zip(axes, a_values):
    x = a**n
    ax.stem(n, x)
    ax.set_title(f"a = {a}")
    ax.set_xlabel("n")
    ax.set_ylabel("Amplitude")

plt.tight_layout()
plt.show()
```

```python
n = np.arange(0, 41)

x1 = np.cos(2*np.pi*n/8)
x2 = np.cos(2*np.pi*n/7.5)

plt.figure(figsize=(10, 4))

plt.stem(n, x1, linefmt="C0-", markerfmt="C0o",
         basefmt=" ", label="cos(2πn/8)")

plt.plot(n, x2, "r--", label="cos(2πn/7.5)")

plt.title("Cosine Signals")
plt.xlabel("n")
plt.ylabel("Amplitude")
plt.legend()
plt.grid()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
import librosa
from IPython.display import Audio, display

# Read your uploaded audio
fname = "/content/My_Voice.m4a"

x, fs = librosa.load(fname, sr=None, mono=True)

# Normalize audio
x = x / (np.max(np.abs(x)) + 1e-10)

# Create time axis
t = np.arange(len(x)) / fs

# Display audio information
print("MY VOICE RECORDING")
print("-----------------------")
print(f"Sampling rate = {fs} Hz")
print(f"Duration = {len(x)/fs:.2f} seconds")
print(f"Number of samples = {len(x)}")

# Plot waveform
plt.figure(figsize=(12, 4))
plt.plot(t, x)
plt.title("My Voice Recording")
plt.xlabel("Time (seconds)")
plt.ylabel("Amplitude")
plt.grid(True)
plt.show()

# Play your voice recording
display(Audio(x, rate=fs))
```
